# Iterative und rekursive Algorithmen

```python
def fib_recursive(n: int) -> int:
    if n <= 2:
        return 1
    else:
        return fib_recursive(n-1)+fib_recursive(n-2)

def fib_iterative(n: int) -> int:
    fib1 = 1
    fib2 = 1
    for _ in range(n-2):
        fib2 += fib1
        fib1 = fib2 - fib1
    return fib2

def fib_recursive_memoized(n: int, cache: dict = {}) -> int:
    if n <= 2:
        return 1
    elif val := cache.get(n, False):
        return val
    else:
        val = fib_recursive_memoized(n-1, cache) + fib_recursive_memoized(n-2, cache)
        cache[n] = val
        return val
```

```python
def comp_and_swap(items: list[int], ind1: int, ind2: int, ascending: bool):
    if (items[ind1] > items[ind2]) == ascending:
        items[ind1], items[ind2] = items[ind2], items[ind1]

def bitonic_merge(items: list[int], low: int, length: int, ascending: bool):
    if length > 1:
        middle = int(length / 2)
        for i in range(low, low + middle):
            comp_and_swap(items, i, i+middle, ascending)
        bitonic_merge(items, low, middle, ascending)
        bitonic_merge(items, low+middle, middle, ascending)

def bitonic_sort(items: list[int], low: int, length: int, ascending: bool):
    if length > 1:
        middle = int(length / 2)
        bitonic_sort(items, low, middle, True)
        bitonic_sort(items, low+middle, middle, False)
        bitonic_merge(items, low, length, ascending)

my_list = [5, 3, 10, 25, 2, 18, 29, 0]
bitonic_sort(my_list, 0, len(my_list), True)
print(my_list) # [0, 2, 3, 5, 10, 18, 25, 29]
```
