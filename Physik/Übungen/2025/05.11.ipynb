### 1.
$$
\begin{align*}
\lambda &\colonequals 590 \text{ nm} \\

\end{align*}
$$

```python
lamb = Rational(590, 1000000000)
e = 2
s_k = Rational(18,200)
k = 1
```

$$
\begin{align*}
\arctan \frac{s_k}{e} &= \arcsin \frac{k\lambda}{b} \\
\sin\arctan \frac{s_k}{e} &= \frac{k\lambda}{b} \\
\rightarrow b &= \frac{k\lambda}{\sin\arctan \frac{s_k}{e}}
\end{align*}
$$

```python
Eq(Symbol("b"), (k*lamb)/sin(atan(s_k/e)).evalf(7)*1000000*Symbol("\\text{µm}"))
```

$$
\begin{align*}
\tan\alpha &\approx \sin\alpha \\
\rightarrow \frac{s_k}{e} &\approx \frac{k\lambda}{b} \\
\rightarrow b &\approx \frac{ek\lambda}{s_k}
\end{align*}
$$

```python
Eq(Symbol("b"), (e*k*lamb/s_k).evalf(7)*1000000*Symbol("\\text{µm}"))
```
