#### a.
Von der Ruhelage ausgehend wird die kinetische Energie der Kugel bei der Auslenkung zu einem Teil in potentielle Energie umgewandelt, bis die kinetische Energie der Kugel null ist. Neben der Umwandlung in potentielle Energie wird gleichzeitig durch die Schwingungsdämpfer ein Teil der kinetischen Energie in thermische Energie umgewandelt und dabei entwertet, da bei der Bewegung der Kolben der Schwingungsdämpfer durch die Auslenkung der Kugel Reibung im Dämpfer entsteht. Am Punkt der Schwingung, an dem die kinetische Energie der Kugel null wird, kehrt sich die Bewegungsrichtung um und die an diesem Punkt maximale potentielle Energie wird wieder in kinetische Energie umgewandelt, so dass sich die Kugel in Richtung ihrer Ruhelage bewegt. Dabei wird erneut ein Teil der kinetischen Energie durch die Reibung in den Schwingungsdämpfern in thermische Energie umgewandelt und damit entwertet. Sobald die Kugel ihre Ruhelage wieder erreicht, wurde die potentielle Energie vollständig in kinetische Energie umgewandelt und die kinetische Energie ist erneut maximal – diesmal jedoch geringer als zum letzten Zeitpunkt, an dem sie sich in ihrer Ruhelage befand, da ein Teil der mechanischen Energie in thermische Energie umgewandelt wurde. Diese Abfolge an Umwandlungen wiederholt sich nun so lange, bis die mechanische Energie vollständig durch die Schwingungsdämpfer entwertet wurde.

#### c.
Der Sinn des Schwingungstilgers ist es, die kinetische Energie des Gebäudes, die die Schwingung hervorruft, möglichst effektiv zu entwerten, damit die Schwingung des Gebäudes möglichst klein bleibt. Dafür ist es sinnvoll, dafür zu sorgen, dass auf den Schwingungstilger möglichst viel der kinetischen Energie des Gebäudes übertragen wird. Da hier von einer erzwungenen Schwingung gesprochen werden kann, da auf den Schwingugnstilger durch das Gebäude eine periodische Kraft ausgeübt wird, lässt sich ableiten, dass die Amplitude des Schwingungstilgers, und damit auch die aufgenommene kinetische Energie, möglichst groß ist, wenn die Eigenfrequenz mit der Frequenz der periodisch ausgeübten Kraft annähernd übereinstimmt. Somit ist der Schwingungstilger am effektivsten, wenn seine Eigenfrequenz mit der des gebäudes annähernd übereinstimmt.

#### d.
$$
\begin{align*}
f_G &= \frac1T \\
&= \frac{46 \frac{\text m}{\text s}}{H_D} \\
f_S &= \frac{\sqrt g}{2\pi\sqrt l}
\end{align*}
$$

```python
f_G = 1/T
f_S = sqrt(g) / (2*pi*sqrt(l))
(
    Eq(Symbol("f_G"), f_G.evalf()),
    Eq(Symbol("f_S"), f_S.evalf()),
    Eq(Symbol("\\Delta f"), (f_G-f_S).evalf()),
    Eq(Symbol("\\frac{\\Delta f}{f_G}"), ((f_G-f_S)/f_G).evalf())
)
```

Die Eigenfrequenzen unterscheiden sich um ca. $\displaystyle 0,026 \text{ Hz} $, was einer Abweichung der Eigenfrequenz des Schwingungstilgers zu der des Gebäudes von rund $\displaystyle 25\% $ entspricht. Da weder $f_G \gg f_S$ noch $f_G \ll f_S$, lässt sich hier von einem annähernden Resonanzfall sprechen.
