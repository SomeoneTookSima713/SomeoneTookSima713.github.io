#### a.
$$
\begin{align*}
f &= \frac{1}{2\pi\sqrt{LC}} \\
L &= \mu_0\mu_r \frac{N^2A_S}{l} \\
C &= \epsilon_0 \epsilon_r \frac{A_K}{d} \\
\rightarrow &= \frac{\sqrt{ld}}{2\pi N\sqrt{\mu_0\epsilon_0 A_S A_K}}
\end{align*}
$$

```python
L = Eval.magnetic_field_constant * N**2 * As / l
C = Eval.epsilon_0 * Ak / d
T = 2*pi*sqrt(L*C)
Eq(Symbol("f"), (1/T).evalf())
```
