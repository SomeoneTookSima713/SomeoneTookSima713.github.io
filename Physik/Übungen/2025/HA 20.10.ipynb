Der reflektierte Strahl ändert seine Richtung abhängig davon, um welche Achse der Spiegel rotiert wird. Wenn er um das Lot der Reflektion rotiert wird, ändert sich die Richtung des reflektierten Strahls nicht. Wenn er um eine zum Lot und dem einfallenden Strahl senkrechte Achse gedreht wird, "dreht" sich der reflektierte Strahl um das Doppelte des Winkels in dieselbe Richtung – außer der Grenzfall, dass der Lichtstrahl in der Ebene des Spiegels liegt, wird erreicht oder überschritten.

$$
\begin{align*}
\frac{h_K}{x} &= \frac{h_S}{s} \\

\end{align*}
$$
