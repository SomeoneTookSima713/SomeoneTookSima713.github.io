```python
alpha = rad(55)
n = Rational(3,2)
d = Rational(5,100)

beta = asin(sin(alpha)/n)

s = sin(pi/2-alpha-beta)*d/cos(beta)

Eq(Symbol("s"), s.evalf())
```
