```python
s = {
    "P_S": 22.5,
    "P_M": 53,
    "P_F": 52,
    "K": 52.5,
    "H_1": 61.5,
    "H_2": 68
}

[Eq(Symbol(f"b_{k}"), (Rational(650,1000000000)/sin(atan(s[k]/9000))).evalf(5)*1000000*Symbol("\\text{µm}")) for k in s]
```

$$
\begin{align*}

\end{align*}
$$
