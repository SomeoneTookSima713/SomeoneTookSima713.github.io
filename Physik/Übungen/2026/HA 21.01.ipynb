### 1.
Das Modell des eindimensionalen Potenzialtopfs mit unendlich hohen Wellen und potenzieller Energie im Topf gleich null soll die möglichen Zustände eines in einem Atom oder Molekül gebundenen Elektrons unter der Annahme, dass die Elektronen mithilfe einer Wellenfunktion über die Schrödinger-Gleichung repräsentiert werden, vereinfacht darstellen. Hierfür wird angenommen, dass sich die Elektronen auf eindimensionalen Bahnen befinden und in ihrer Bewegung auf die Wände des "Potentialtopfes" eingeschränkt sind. Damit die Elektronen eine stabile Position einnehmen können und nicht ihre ganze kinetische Energie abstrahlen und in den Kern fallen, müssen die Wellenfunktionen jedes Elektrons nun eine Stehende Welle innerhalb des Topfes bilden, wodurch nur bestimmte kinetische Energien für ein im Atom gebundenes Elektron zugelassen werden.

Im Modell des Potenzialtopfs wird außerdem die kinetische Energie des Elektrons, da es eine stehende Welle ist und sich demnach, bis auf durch die Effekte der Heisenberg'schen Unbestimmtheitsrelation, nicht wirklich im Raum bewegt, als potenzielle Energie in Form der Lokalisationsenergie aufgefasst: Das Nullniveau dieser potenziellen Energie wird bei einer Lokalisationsenergie von null gesetzt und erstreckt sich, zur Vereinfachung der Modellvorstellung, ins Unendliche.

Bei der Absorption eines Photons durch ein Atom nimmt eines der gebundenen Elektronen die Energie des Photons auf. Dadurch, dass die Wellenfunktion des Elektrons allerdings eine stehende Welle im Potentialtopf bilden muss, kann das Elektron nur bestimmte, diskrete Energiewerte annehmen. Demnach kann es auch nur bestimmte Energiemengen aufnehmen, da sonst keines dieser Energieniveaus erreicht werden kann, wodurch nur Photonen bestimmter Wellenlängen absorbiert werden.

$$
\begin{align*}
\lambda_1 &= 121,5 \text{ nm} \\
E_n &= \frac{h^2}{8m_eL^2}n^2 \\
\Delta E(n\to m) &= E_m - E_n \\
&= \frac{h^2}{8m_eL^2}(m^2-n^2) \\
&= \frac{hc}{\lambda_1} \\
L^2 &= \frac{h^2\lambda_1}{8m_ehc}(m^2-n^2)
\end{align*}
$$

Da $\lambda_1$ die größte von Wasserstoff absorbierte Wellenlänge und damit das Licht mit der niedrigsten Energie ist, kann ein Übergang von Energieniveau 1 zu 2 bei der Absorption angenommen werden. Demnach kann $n=1$ und $m=2$ angenommen werden:

$$
\begin{align*}
L^2 &= \frac{h^2\lambda_1}{8m_ehc}(2^2-1^2) \\
\Rightarrow L &= \sqrt{\frac38\frac{h\lambda_1}{m_ec}}
\end{align*}
$$

```python
L = sqrt(Rational(3,8)*Eval.planck_constant*121.5e-9/Eval.mass.electron/Eval.light_speed)
Eq(Symbol("L"), L.evalf()*Symbol("\\text{m}"))
```

\end{align*}
$$

### 3.
$$
\begin{align*}
E_1(n) &= \frac{h^2}{8m_eL^2}n^2 \\
E_2(n) &= -13,61 \text{ eV} \cdot \frac{1}{n^2} \\
E &= \frac{hc}{\lambda} \\ \\
\rightarrow \lambda_n &= \frac{hc}{E(n+1)-E(1)} \\
\rightarrow \lambda_{n;1} &= \frac{hc}{E_1(n+1)-E_1(1)} \\
\rightarrow \lambda_{n;2} &= \frac{hc}{E_2(n+1)-E_1(1)} \\
\end{align*}
$$

```python
n = Symbol("n")
E1 = Eval.planck_constant**2/8/Eval.mass.electron/L**2*n**2
E2 = -13.61 / n**2 * Eval.elementary_charge

lamb1 = Eval.planck_constant * Eval.light_speed / (E1.subs(n, n+1) - E1.subs(n,1))
lamb2 = Eval.planck_constant * Eval.light_speed / (E2.subs(n, n+1) - E2.subs(n,1))
```

```python
[Eq(Symbol(f"\\lambda_{{{i};1}}"), lamb1.subs(n, i) * Symbol("\\text{ m}")) for i in range(2, 4)]
```

```python
[Eq(Symbol(f"\\lambda_{{{i};2}}"), lamb2.subs(n, i) * Symbol("\\text{ m}")) for i in range(2, 4)]
```
