### Nr. 11.

$$
\begin{align*}
b(t) &= 10^6 \frac{\text t}{\text a} \cdot e^{-0,1 \text{ a}^{-1} \cdot t} \\
t &\in [0 \text{ a};10 \text{ a}] \\
B(t) &= \int b(t) dt \\
&= -10^7 \text{ t} \cdot e^{-0,1 \text{ a}^{-1} \cdot t} \\
&= - 10^7 \text{ t} \cdot e^{-0,1 \text{ a}^{-1} \cdot t} \\
4.000.000 \text{ t} + \left[B(t)\right]_{0 \text{ a}}^{10 \text{ a}} &= 4.000.000 \text{ t} + B(10 \text{ a}) - B(0 \text{ a}) \\
\end{align*}
$$

```python
def B(t):
    return - Integer(10)**7 * E**(Rational(-1,10)*t)

Eq(Symbol("4.000.000 \\text{ t} + [B(t)]_{0 \\text{ a}}^{10 \\text{ a}}"), (Integer(4000000) + B(10) - B(0)).evalf())
```
