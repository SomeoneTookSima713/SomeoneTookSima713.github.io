Die Fläche zwischen dem Graphen von $f$ mit $f(x) = e^{-x}$ und der $x$-Achse über dem Intervall $[1;2]$ rotiert um die $x$-Achse. Skizzieren Sie den entstehenden Rotationskörper und berechnen Sie das Volumen.

```python
# -------------------------------------------------------
# 1. Funktion definieren
# -------------------------------------------------------
def f(x):
    return 2.71828**(-x)   # Beispiel-Funktion

# -------------------------------------------------------
# 2. Intervall und Diskretisierung
# -------------------------------------------------------
a, b = 1, 2            # Definitionsintervall
n = 200                 # Anzahl der x-Punkte
m = 100                 # Anzahl der Winkelpunkte (für Rotation)

x = np.linspace(a, b, n)
theta = np.linspace(0, 2*np.pi, m)

# 2D-Gitter für die 3D-Koordinaten erzeugen
X, THETA = np.meshgrid(x, theta)

# Radius entspricht f(x)
R = f(X)

# -------------------------------------------------------
# 3. Rotationskörper (Rotation um x-Achse)
# -------------------------------------------------------
Y = R * np.cos(THETA)
Z = R * np.sin(THETA)

# -------------------------------------------------------
# 4. Plot
# -------------------------------------------------------
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.85)

# Achsenbeschriftung
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title("Rotationskörper um die x-Achse")

plt.show()
```
