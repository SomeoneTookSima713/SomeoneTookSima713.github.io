#### a.
$$
\begin{align*}
y &= x^2 - 1 \\
\rightarrow A &= \int_{-2}^0 \left| x^2 - 1 \right|dx \\
&= \int_{-2}^0 \left( \begin{cases} x^2-1 & x \notin (-1;1) \\-x^2+1 & x \in(-1;1) \end{cases} \right)dx \\
&= \left[ \begin{cases} \frac13x^3 - x & x \notin (-1;1) \\-\frac13x^3 + x & x \in (-1;1) \end{cases} \right]_{-2}^0 \\
&= \left[ \frac13x^3 - x \right]_{-2}^{-1} + \left[ -\frac13x^3 + x \right]_{-1}^0 \\
&= \frac23+\frac23 + \frac23 \\
&= 2
\end{align*}
$$

$$
