### Berechnen Sie die Gleichung der Stammfunktion von $f$, dessen Graph durch den Punkt $P$ verläuft.
#### a. $f(x) = 4x + 3;~ P(-4|15)$

$$
\begin{align*}
F(x) &= \int f(x)dx \\
&= 2x^2 + 3x + c \\
F(-4) &\overset!= 15 \\
\rightarrow 15 &= 2\cdot(-4)^2 + 3\cdot(-4) + c \\
\rightarrow c &= -5 \\
\Rightarrow F(x) &= 2x^2 + 3x - 5
\end{align*}
$$

#### b. $f(x) = 6x^2 - 2x + 1;~ P(1|3)$

$$
\begin{align*}
F(x) &= \int f(x)dx \\
&= 2x^3 - x^2 + x + c \\
F(1) &\overset!= 3 \\
\rightarrow c &= 1 \\
\Rightarrow F(x) &= 2x^3 - x^2 + x + 1
\end{align*}
$$
