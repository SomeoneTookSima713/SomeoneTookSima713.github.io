$A :$ Der erste Würfel zeigt eine 6.

$B_a :$ Die Summe lautet 7.

$B_b :$ Die Summe lautet 8.

$$
\begin{align*}
\Omega &= \set{1;2;3;4;5;6}^2 \\
A &= \set{(6;b)|(6;b)\in\Omega} \\
B_a &= \set{(a;b)|(a;b)\in\Omega, a+b = 7} \\
B_b &= \set{(a;b)|(a;b)\in\Omega, a+b = 8} \\
\end{align*}
$$

#### a.
$$
\begin{align*}
P(A) &= \frac16 \\
P(B) &= \frac16 \\
P(A\cap B) &= P(\set{(6;b)|(6;b)\in\Omega, 6+b=7}) \\
&= P((6;1)) \\
&= \frac1{36} \\
&= P(A) \cdot P(B)
\end{align*}
$$

#### b.
$$
\begin{align*}
P(A) &= \frac16 \\

&= P((6;2)) \\
&= \frac1{36} \\

\end{align*}
