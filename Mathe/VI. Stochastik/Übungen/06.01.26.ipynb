Produktion von Schrauben.

3 Maschinen $M_1$, $M_2$ und $M_3$

$20\%$ der Gesamtprodukte durch $M_1$, Ausschuss: $1\%$ \
$50\%$ der Gesamtprodukte durch $M_2$, Ausschuss: $4\%$ \
$30\%$ der Gesamtprodukte durch $M_3$, Ausschuss: $5\%$

Berechnen Sie die Wahrscheinlichkeit, dass eine zufällig entnommene Schraube Ausschuss ist.

$$
\begin{align*}
P(A) &= 20\% \cdot 1\% + 50\% \cdot 4\% + 30\% \cdot 5\% \\
&= 3,7\%
\end{align*}
$$

Interpretieren Sie $P(M_1|A)$ im Sachzusammenhang und berechnen Sie $P(M_1|A)$, $P(M_2|A)$, $P(M_3|A)$.

$P(M_1|A)$ beschreibt die Wahrscheinlichkeit, dass eine Schraube durch die Maschine $M_1$ produziert wurde, wenn sie Ausschuss ist.

$$
\begin{align*}
P(M_1|A) &= \frac{P(M_1 \cap A)}{P(A)} \\
&= \frac{0,2\%}{3,7 \%} \\
&= \frac{2}{37} \\
P(M_2|A) &= \frac{20}{37} \\
P(M_3|A) &= \frac{15}{37}
\end{align*}
$$
