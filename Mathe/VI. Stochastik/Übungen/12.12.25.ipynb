### Nr. 2.
$$
\begin{align*}
K &:= \set{a_1;a_2;a_3;h_1;h_2} \\
K_a &:= \set{a_1;a_2;a_3} \\
K_h &:= \set{h_1;h_2} \\
\exists! \Omega &: a, b \in K ~\forall (a;b)\in\Omega \\
E &= \set{(a; b) | (a; b) \in \Omega,\, a \in K_a} \\
F &= \set{(a; b) | (a; b) \in \Omega,\, b \in K_h} 
\end{align*}
$$

#### a.
$$
\begin{align*}
\Omega &= K^2 \\
P(E) &= \frac35 \\
P(F) &= \frac25 \\
P(E\cap F) &= P(\set{(a;b)|(a;b) \in \Omega,\, a \in K_a,\, b \in K_h}) \\
&= P(K_a \times K_h) \\
&= \frac 35 \cdot \frac 25 \\
&= P(E) \cdot P(F)
\end{align*}
$$

#### b.
$$
\begin{align*}
\Omega &= \bigcup_{k\in K} \set{k} \times (K\setminus k) \\
P(E) &= \frac35 \\
P(F) &= \frac35 \cdot \frac12 + \frac25 \cdot \frac14 \\
&= \frac25 \\
P(E\cap F) &= \frac 35 \cdot \frac12 \\
&\neq P(E) \cdot P(F) \\
P_E(F) &= \frac12
\end{align*}
$$
