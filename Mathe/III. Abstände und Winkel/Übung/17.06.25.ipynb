### Aufgabe von Frau Burnett
Gegeben sind die Ebenen $E_k$ mit $\displaystyle E_k: x + (k-1)y + (k+1)z = 5 ~\,(k\in\mathbb R) $ und die Gerade $g$, die durch die Punkte $A(-4|5|4)$ und $B(-3|7|2)$ verläuft.
Berechnen Sie $k$ so, dass sich $E_k$ und $g$ in einem Winkel von $30\degree$ schneiden.

$$
\begin{align*}
g: \overrightarrow{ x } &= \begin{pmatrix}-4\newline 5\newline 4\end{pmatrix} + t \begin{pmatrix}1\newline 2\newline -2\end{pmatrix} \\
\overrightarrow{ v_n } &= \begin{pmatrix}\frac13\newline \frac23\newline -\frac23\end{pmatrix} \\
\sin30\degree &= \frac{ \begin{pmatrix}1\newline k-1\newline k+1\end{pmatrix} \circ \overrightarrow{ v_n } }{ \left| \begin{pmatrix}1\newline k-1\newline k+1\end{pmatrix} \right| } \\
\end{align*}
$$

```python
k = symbols('k', real=True)
vn = Vec3(1,2,-2).unit
tuple([Eq(Symbol('k_'+str(i)), sol) for i, sol in enumerate(solve((abs(Vec3(1,k-1,k+1).unit * vn) - sin(pi/6)).simplify(), k))])
```
