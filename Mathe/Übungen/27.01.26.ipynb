## IQB B-Teil Analysis 2025 (S. 108 in der PDF)

### 1.
$$ f_k(x) := \frac 1{2k}x^2(x-2k)^2~,\,k\in\mathbb R^+ $$

```python
k, x = symbols("k x", real=True)

f = 1/(k*2)*x**2*(x-2*k)**2
```

#### a.
Es lässt sich beweisen, dass $f_k(x)$ für alle $k\in\mathbb R^+$ genau zwei Nullstellen existieren, indem man $f_k(x) = 0$ nach $x$ umstellt und in beide Lösungen, die dabei rauskommen, jeder Wert von $k$ eingesetzt werden kann:

$$
\begin{align*}
f_k(x) &= 0 \\
\frac1{2k}x^2(x-2k)^2 &= 0 \\
\frac1{2k}x^2(x^2 - 4kx + 4k^2) &= 0 \\
\frac1{2}x^2\left(\frac1kx^2 - 4x + 4k\right) &= 0 \\
\rightarrow x=0 &\in \mathcal L \\
x^2 - 4kx + 4k^2 &= 0 \\
x &= 2k \pm \sqrt{4k^2-4k^2} \\
&= 2k \\
\Rightarrow \mathcal L &= \set{0; 2k} \\
\Rightarrow \left| \mathcal L \right| &= 2
\end{align*}
$$

Die Nullstellen für $f_k(x)$ sind $x_1=0$ und $x_2 = 2k$.

#### b.
$$
\begin{align*}
f_k'(x) &= 0
\end{align*}
$$

```python
sols_b_diff = solveset(f.diff(x), x)
sols_b_diff
```

Für jede der Lösungen der obigen Gleichung wird nun $f''_k(x)$ berechnet. Ist der resultierende Wert positiv, handelt es sich um einen Tiefpunkt. Ist der Wert negativ, handelt es sich um einen Hochpunkt.

```python
{Symbol(f"f''_{{k}}({s})"): f.diff(x,x).subs(x, s).simplify() for s in sols_b_diff.args}
```

$$
\begin{align*}
T_1&(0|f_k(0)) \\
T_2&(2k|f_k(2k)) \\
H&(k|f_k(k))
\end{align*}
$$

Die Distanz zwischen dem Hochpunkt und den Tiefpunkten lässt sich nun mit dem Satz des Pythagoras über die Differenzen der $x$- und $y$-Koordinaten der Punkte berechnen:

$$
\begin{align*}
d &= \sqrt{(x_H-x_{T_1})^2 + (y_H-y_{T_1})^2} \\
&= \sqrt{(x_H-x_{T_2})^2 + (y_H-y_{T_2})^2} \\
\end{align*}
$$

```python
Eq(Symbol("d"), sqrt(k**2+(f.subs(x,k)-f.subs(x,0))**2))
```

#### c.
Damit dieser Integral den Flächeninhalt berechnet, muss für jedes $x\in[-1;1]$ $f_k(x)$ dasselbe Vorzeichen haben. Dementsprechend lässt sich die Aussage mittels der Lösung der Gleichung $f_k(x)<0$ be- oder widerlegen (unter der Annahme, dass alle Funktionswerte im Intervall $[-1;1]$ positiv sein werden):

$$
\begin{align*}
f_k(x) < 0 \\
\frac 1{2k}x^2(x-2k)^2 < 0 \\
x^2\frac{x^2 - 4kx + 4k^2}{2k} < 0 \\
x^2\left(x-2k\right)^2 < 0 \\
\end{align*}
$$

Wie zu erkennen ist, existiert keine Lösung für $x$, die diese Gleichung erfüllt, da $x^2$ sowie $(x-2k)^2$ immer positiv sind.
