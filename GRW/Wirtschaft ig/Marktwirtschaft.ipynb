# Angebot und Nachfrage auf einem Markt

```python
def nachfrage(m):
    return 0.1/m

def angebot(m):
    return 0.1/(1-m)

plt.plot([i/40 for i in range(1,41)], [nachfrage(i/40) for i in range(1,41)])
plt.plot([i/40 for i in range(0,40)], [angebot(i/40) for i in range(0,40)])
plt.ylabel("Preis P")
plt.xlabel("Menge m")
plt.axis((0, 1, 0, 1))
plt.xticks([])
plt.yticks([])
plt.plot(0.5, nachfrage(0.5), "rx", label = "Gleichgewichtspreis")
plt.plot(0.5, 0.5, "rx", label = "Cournotscher Punkt")
plt.annotate("Gleichgewichtspreis", (0.5, nachfrage(0.5)))
plt.annotate("Cournotscher Punkt", (0.5, 0.5))
plt.grid(true, "major", "both", color='#888888', linestyle='--', linewidth=0.5)
plt.show()
```

Die blaue Kurve ist die der Nachfrage, die orangene die des Angebots. Der Schnittpunkt ist der Gleichgewichtspreis. Oberhalb beider Kurven existiert ein Angebotsüberschuss (es gibt für alle Nachfrager genug, aber der Preis ist zu hoch) und unterhalb der Kurven ein Nachfrageüberschuss (es gibt nicht für alle Nachfrager genug).

Wenn es sehr viele Anbieter und Nachfrager gibt, spricht man von einem *Polypol*. Wenn es nur auf einer Seite viele gibt (z.B. viele Nachfrager und wenige Anbieter), spricht man von einem *Oligopol*.

Wenn es nur einen Nachfrager oder nur einen Anbieter gibt, spricht man von einem *Monopol*. Hierbei gibt es einen Monopolprofit, der Preis beim Gleichgewichtspunkt liegt also über dem Gleichgewichtspreis. Dieser Punkt $(m_G; P_G + P_m)$, wobei $m_G$ udn $P_G$ die Koordinaten des Gleichgewichtspunktes sind und $P_m$ der Monopolprofit ist, nennt sich *Cournotscher Punkt*.

#### Merkmale
* eine Preisveränderung ist annähernd indirekt proportional zur Nachfrage: steigt der Preis, sinkt die Nachfrage bedeutsam
    * das passiert nur, wenn es ein Substitutionsgut gibt (einen gleichwertigen, gleich guten Ersatz → *nicht für jedes Gut gibt es ein Substitutionsgut*)
* eine Verschiebung der Kurve auf der x-Achse (Mengenachse) kann aus ein paar üblichen Gründen auftreten:
    * saisonale Verschiebungen
    * Veränderungen im Nachfrageverhalten (z.B. Überspringen auf ein Substitutionsgut → Nachfrage des Substitutionsgutes steigt unabhängig vom Preis)
    * Entstehung von Gerüchten, Qualitätsmängeln, Rückzugsaktionen, ...

    → der Markt ist stark von spontanen äußeren Einflüssen geprägt, *volatil*

## Der Konjunkturverlauf
###### Verlauf von Angebot und Nachfrage in der gesamten Volkswirtschaft

```python
plt.plot([i/40 for i in range(0, 41)], [sin(2*pi*i/40).evalf() for i in range(0, 41)])
plt.ylabel("Intensität I")
plt.xlabel("Zeit t")
plt.axis((0, 1, -1, 1))
plt.xticks([])
plt.yticks([])
plt.annotate("1", (0.125+0.015, sin(0.25*pi)-0.015))
plt.annotate("2", (0.25, sin(0.5*pi)-0.1))
plt.annotate("3", (0.375+0.015, sin(0.75*pi)-0.015))
plt.annotate("4", (0.75, sin(1.5*pi)+0.1))
plt.grid(true, "major", "both", color='#888888', linestyle='--', linewidth=0.5)
plt.show()
```

Phasen der Konjunktur:
1. Aufschwung
2. Höhepunkt
3. Rezession
4. Depression

Indikatoren:
* Produktivität
* Preise
* Löhne
* Arbeitskräftepotenzial
* Aktien

#### Indikatoren im Aufschwung
* Produktivität nimmt zu
* Preise nehmen zu
* Löhne nehmen zu
* Arbeitskräfte werden eingestellt
* Aktien gewinnen an Wert

#### Indikatoren im Höhepunkt
* *Marktsättigung*
    * Angebot und Nachfrage sind vollständig befriedigt
* durch Sättigung sinkt Nachfrage → Preise sinken → Bewegung in eine Rezessionsphase

#### Indikatoren in der Rezession
* Produktivität sinkt (weil Angebot, als Reaktion auf Lagerhaltung durch weniger Nachfrage, sinkt)
* Arbeitskräfte werden entlassen
* Preise sinken
* Löhne sinken (theoretisch)
* Aktien verlieren an Wert

#### Indikatoren in der Depression
* Produktivität minimal
* Preise minimal
* Löhne minimal
* relativ hohe Arbeitslosigkeit
* Aktien "sind im Keller"
* durch Innovation kommt man wieder in eine Aufschwungphase

**Das ist in der Marktwirtschaft so. In der sozialen Marktwirtschaft funktioniert das zwar größtenteils auch, aber hier hat man Tarifverträge, Entlassungen sind schwieriger → Probleme für Konjunkturkurve.**

### Merkmale und Maßnahmen zur Konjunktursteuerung von Friedman und Keynes

#### Friedman
##### Merkmale
* reduzierte Einmischung der Regierung in die Wirtschaft
    * immer noch bedeutende Rolle, z.B. Lieferung eines festen Geld- und Finanzrahmens
* es werden der Regierung keine Funktionen zugeteilt, die der Markt ausführen kann

##### Maßnahmen
* Bereitstellung legaler und ökonomischer Rahmen, der einzelnen Menschen befähigt, wirtschaftliches Wachstum zu produzieren
* Regierung trägt Verantwortung für Geldwesen
    * in einem kontrollierten Rahmen, zur Verhinderung der Schwächung der freien Gesellschaft
    * dafür müssen Regeln zur Führung der Geldpolitik gesetzlich festegelegt werden

#### Keynes
##### Merkmale
* nimmt Weltwirtschaftskrise von 1930er-Jahren als Ausgangssituation an
* Anzahl der Arbeiter wird durch Reallohn bestimmt
    * Reallohn ist Lohn relativ zu Preisen von Gütern
* in Rezession sinken Preise oft schneller als Lohnniveau

    → Reallöhne steigen; mehr Leute wollen Arbeiten, weniger werden durch Lagerhaltung gebraucht \
    → Arbeitslosigkeit steigt
* Möglichkeit zur Eliminierung Arbeitslosigkeit: Durck auf Nominallöhne durch Arbeitslose ausüben (sind dazu bereit, zu weit geringeren Löhnen zu arbeiten)
* nach klassischer Ökonomie reicht das, um Reallöhne zu senken und Arbeitslosigkeit zu mindern
    * nach Keynes verhindern starre Nominallöhne aber dann solche Anpassung → Andauern unfreiwilliger Arbeitslosigkeit, Unternehmen und Arbeiter Gefangen in Kreislauf von Arbeitslosigkeit und Unterproduktion

##### Maßnahmen
* Fiskalpolitik:
    * in Rezession Staatsausgaben erhöhen, um Nachfragelücke zu füllen
    * in Aufschwung Steuern erhöhen, um Nachfrage zu bremsen
* Geldpolitik:
    * in Rezession Zinsen senken, um Investitionen anzuregen
