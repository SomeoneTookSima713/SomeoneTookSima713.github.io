#### b.
Damit die Photonen mit $2,55 \text{ eV}$ im sichtbaren Bereich liegen, muss ihre Wellenlänge zwischen $330 \text{ nm}$ und $780 \text{ nm}$ liegen:

$$
\begin{align*}
330 \text{ nm} &< \lambda < 780 \text{ nm} \\
330 \text{ nm} &< \frac cf < 780 \text{ nm} \\
330 \text{ nm} &< \frac {hc}E < 780 \text{ nm} \\
\frac{hc}{E} &= \frac{h \cdot c}{2,55 \text{ eV}} \\
&\approx 4,86\cdot10^{-7} \text{ m} \\
&\approx 486 \text{ nm} \\
330 \text{ nm} &< 486 \text{ nm} < 780 \text{ nm} ~~\text{w.A.}
\end{align*}
$$

#### c.
Beim Vorgang der Emission von Licht wird von einer Ladung Energie in Form einer elektromagnetischen Welle abgegeben. Dies passiert oftmals mit Elektronen, die in einem Atom gebunden sind, wenn sie von einem höheren auf ein niedrigeres Energieniveau fallen. Dadurch sinkt ihre kinetische Energie und ein Photon wird mit der Energiedifferenz erzeugt.

Atomspektren sind Linienspektren, weil Atome bei der Änderung des Energieniveaus eines Elektrons von einem höheren auf ein niedrigeres Photonen aussenden, diese Energieniveaus aber nur für bestimmte, diskrete Werte der kinetischen Energie des Elektrons existieren. Deshalb gibt es auch nur bestimmte, diskrete Energiedifferenzen, mit denen Photonen ausgesendet werden können, weshalb es auch nur diskrete Wellenlängen gibt, mit denen Photonen ausgesendet werden.
