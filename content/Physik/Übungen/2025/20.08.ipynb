```python
alpha, t, m = symbols('alpha t m', real=True)
```

$$
\begin{align*}
E_{\text{kin}1} &= E_{\text{pot}0} \\
\frac12mv^2 &= mgh \\
\rightarrow v &= \sqrt{2gh} \\
a_H &= g\sin(\alpha) \\
&= \frac{v}{t_1} \\
\rightarrow t_1 &= \frac{\sqrt{2gh}}{g\sin(\alpha)}
\end{align*}
$$
$t_1$ entspricht dem Zeitpunkt, an dem der Ball seinen tiefsten Punkt erreicht. Da die ganze periodische Bewegung symmetrisch ist, entspricht die Periodendauer $T$ dem Vierfachen von $t_1$:
$$
\begin{align*}
T &= 4t_1 \\
&= \frac{4\sqrt{2gh}}{g\sin(\alpha)}
\end{align*}
$$
