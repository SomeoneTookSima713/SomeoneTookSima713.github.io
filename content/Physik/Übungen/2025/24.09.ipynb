### 2.

$$
\begin{align*}
m &\in \mathbb R \\
D_1, D_2 &\in \mathbb R \\
\end{align*}
$$

Es wird angenommen, dass das Koordinatensystem so ausgerichtet ist, dass eine Auslenkung in positiver x-Richtung einer Bewegung im Diagramm nach rechts entspricht.

#### a.
$$
\begin{align*}
F_1(x) &= -D_1x \\
F_2(x) &= -D_2x \\
F_\text{res}(x) &= F_1(x) + F_2(x) \\
&= -(D_1 + D_2)x
\end{align*}
$$

#### b.
$$
\begin{align*}
F_\text{res} &= m \cdot a \\
\rightarrow a &= \frac{F_\text{res}}{m} \\
&= \frac{-(D_1 + D_2)x}{m} \\
\ddot x &= \frac{-(D_1 + D_2)}{m}x \\
\rightarrow 0 &= m\ddot x + (D_1 + D_2)x \\
&= m\ddot x + D_\text{eff}x \\
\rightarrow D_\text{eff} &= D_1 + D_2
\end{align*}
$$

#### c.
$$
\begin{align*}
T &= 2\pi\sqrt{\frac{m}{D}} ~~~~|~D \colonequals D_\text{eff} \\
&= 2\pi\sqrt{\frac m{D_1 + D_2}}
\end{align*}
$$
