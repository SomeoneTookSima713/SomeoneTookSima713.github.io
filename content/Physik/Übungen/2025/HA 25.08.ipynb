## 1.
$$
\begin{align*}
m &\colonequals 200 \text{ g} \\
h_0 &\colonequals 60 \text{ cm} \\
T &\colonequals 0,80 \text{ s}
\end{align*}
$$

```python
m = Rational(2, 10) # kg
h0 = Rational(6, 10) # m
T = Rational(8, 10) # s
```

### a.
$$
\begin{align*}
T &= 2\pi\sqrt\frac mD \\
\rightarrow D &= \frac{4\pi^2m}{T^2} \\
\end{align*}
$$

```python
D = 4*pi**2*m/(T**2) # N/m

Eq(Symbol('D'), D.evalf())
```

### b.
$$
\begin{align*}
F_\text{Spann} &= F_G \\
-Ds &= mg \\
\rightarrow s &= -\frac{mg}D \\
h &= h_0 + s \\
&= h_0 - \frac{mg}D
\end{align*}
$$

```python
h_b = h0 - m*Rational(981, 100)/D
Eq(Symbol('h'), h_b.evalf())
```

### c.
$$
\begin{align*}
y_\text{max} &= h_0 - h \\
y(t) &= y_\text{max}\cos\left( \frac{2\pi}{T}t \right) + h
\end{align*}
$$

```python
t = symbols('t', real=True)
y = Function('y')(t)
y_max = h0-h_b

eq_c = y_max * cos(2*pi/T * t) + h_b

Eq(y, eq_c)
```

```python
expr_c = lambdify(t, eq_c, "math")
y_vals_c = []
for f in range(101):
    y_vals_c.append(float(expr_c(f/50)))
plt.plot([f/50 for f in range(101)], y_vals_c)
plt.ylabel("y in Metern")
plt.xlabel("t in Sekunden")
plt.axis((0, 2, 0, 0.8))
plt.grid(true, "major", "both", color='#888888', linestyle='--', linewidth=0.5)
plt.show()
```
