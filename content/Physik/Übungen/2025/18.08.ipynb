### d.
$$
\begin{align*}
v(t) &= \dot x ~,\,t\text{ in s},\,v\text{ in } \frac{\text cm}{\text s} \\
&= 2\pi\cos(\frac\pi2 x) \\
v(0 \text{ s}) &= 2\pi\cos(0) \\
&= 2\pi \frac{\text{cm}}{\text s} \\
&= 0,02\pi \frac{\text m}{\text s}
\\~\\
E_\text{kin} &= \frac12mv^2 \\
&= \frac12\cdot 0,2 \text{ kg} \cdot \left(0,02\pi \frac{\text m}{\text s}\right)^2 \\
&= 0,00004\pi^2 \text{ J} \\
&= 0,04\pi^2 \text{ mJ}
\end{align*}
$$

```python
Eq(Symbol("E_\\text{kin}"), (0.04*pi**2).evalf())
```
