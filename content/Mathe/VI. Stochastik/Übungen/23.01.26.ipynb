### Nr. 2.
$$

\end{align*}

#### a.
$$
\begin{align*}
P(X = 0) &\leq 0,05 \\
\binom n0 p^0 (1-p)^{n-0} &\leq 0,05 \\
(1-p)^n &\leq 0,05 \\
n\ln(1-p) &\leq \ln(0,05) \\
n &\geq \frac{\ln 0,05}{\ln(1-p)} \\
n &= \left\lceil\frac{\ln 0,05}{\ln(0,75)}\right\rceil \\
\end{align*}
$$

```python
Eq(Symbol("n"), ceiling(ln(0.05)/ln(0.75)).evalf())
```

#### b.
$$
\begin{align*}
P(X\leq 5) &\leq 0,5 \\
\sum_{i=0}^5 \binom ni p^i (1-p)^{n-i} &\leq 0,5
\end{align*}
$$

```python
n, i, p = symbols("n i p", real=True)

sol = nsolve(Eq(Sum(binomial(n, i)*p**i*(1-p)**(n-i), [i, 0, 5]).subs(p, 0.25).doit(), 0.5), n, 11)
Eq(Symbol("n"), ceiling(sol).evalf())
```

#### c.
$$
\begin{align*}
P(X\geq 5) &\geq 0,5 \\
\sum_{i=5}^n \binom ni p^i (1-p)^{n-i} &\geq 0,5
\end{align*}
$$

```python
def nsolve_ineq(equation, expected_val, var, start, step, max_iters=1000):
    curr = start
    for _ in range(max_iters):
        val = equation.subs(var, curr).evalf()
        if (step > 0 and val >= expected_val) or (step < 0 and val <= expected_val):
            return curr
        curr += step
    raise RuntimeError("Couldn't find solution in given iteration count!")

# sol = nsolve(Eq(Sum(binomial(n, i)*p**i*(1-p)**(n-i), [i, 5, n]).subs(p, 0.25), 0.5), n, 11)
sol = nsolve_ineq(
    Sum(binomial(n, i)*p**i*(1-p)**(n-i), [i, 5, n]).subs(p, 0.25),
    0.5,
    n,
    11,
    1
)
Eq(Symbol("n"), ceiling(sol).evalf())
```
