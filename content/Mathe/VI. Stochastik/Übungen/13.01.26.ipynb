### 3.
$\displaystyle X - \text{Gewürfelte Augenzahl} $

$$
\begin{align*}
\exists a,b,c &\in \set{3;4;5;6} \\
E(X) &= \frac16\left( 1 + 5 + 5 + a + b + c \right) \\
&= 4 \\
\rightarrow a+b+c &= 13 \\
\end{align*}
$$

Es gibt zwei Möglichkeiten für die Werte von $a,b,c$: Entweder sie sind alle gleich und nicht $5$, damit die sechs Seiten des Würfels insgesamt drei Zahlen enthalten, oder zwei von ihnen sind gleich $5$ und eine ungleich $5$.

Wenn man annimmt, dass alle drei Zahlen gleich sind, müssten Sie die folgende Gleichung erfüllen:
$$
\begin{align*}
a + b + c &= 13 \\
a &= b = c \\
\rightarrow 3a &= 13
\end{align*}
$$

Da allerdings alle drei Werte natürliche Zahlen sind, kann diese Gleichung nicht erfüllt werden, da hierfür $3$ ein Teiler von $13$ sein müsste. Demnach müssen zwei der Zahlen $a, b, c$ gleich $5$ sein und die andere in der Menge $\set{3;4;6}$ enthalten sein.

Es ist irrelevant, in welcher Reihenfolge die Werte der Variablen gesetzt werden, da das Würfeln einer jeden Seite gleich groß ist. Demnach kann angenommen werden, dass $\displaystyle a=b=5 $ udn nur noch für $c$ gelöst werden:

$$
\begin{align*}
a + b + c &= 13 \\
5 + 5 + c &= 13 \\
\rightarrow c &= 3
\end{align*}
$$

Demnach existiert die Lösung $\set{a=5; b=5; c=3}$, die die beiden ersten Bedingungen erfüllt. Nun muss die letzte Bedingung noch auf Richtigkeit geprüft werden.

Die Wahrscheinlichkeit, dass man zwei gleiche Augenzahlen hintereinander würfelt, setzt sich aus drei Ereignissen zusammen:
* zwei Einsen,
* zwei Fünfen und
* zwei Dreien.

Die insgesamte Wahrscheinlichkeit für zwei Gleiche lässt sich also aus der Summe der Wahrscheinlichkeiten dieser drei Ereignisse berechnen:

$$
\begin{align*}
P(\text{zwei Gleiche}) &= \frac 16 \cdot \frac 16 + \frac46 \cdot \frac 46 + \frac16 \cdot \frac 16 \\
&= \frac1{18} + \frac8{18} \\
&= \frac9{18} \\
&= \frac12
\end{align*}
$$

Hiermit sind alle drei Bedingungen mit der Lösung $\displaystyle \set{a=5;b=5;c=3} $ erfüllt, was bestätigt, dass sich die nicht sichtbaren Seiten des Würfels so beschriften lassen, dass er alle drei Eigenschaften besitzt.
