$$
\begin{align*}
p: y &= -(x-2)^2 + 4 \\
&= -x^2 + 4x \\
g: y &= mx \\
y_p &= y_g  &&|~z \colonequals x,\,z>0\\
-z^2 + 4z &= mz \\
\rightarrow 0 &= z(-z+4-m) \\
\rightarrow z &= 4-m
\end{align*}
$$

$$
\begin{align*}
A_\text{rot} &= \int_0^z (y_p-y_g)dx \\
&= \left[ -\frac13x^3 + x^2\left(2-\frac12m\right) \right]_0^z \\
&= -\frac13z^3 + z^2\left(2 - \frac12m\right) \\
&= -\frac16(m-4)^3 \\
A_\text{blau} &= \int_z^4 (y_g - y_p)dx \\
&= \left[ \frac13x^3 + x^2\left( \frac12m-2 \right) \right]_z^4 \\
&= 8m - \frac13z^3 - z^2\left( \frac12m - 2 \right) - \frac{32}3 \\
&= \frac16m^2(12-m)
\end{align*}
$$
