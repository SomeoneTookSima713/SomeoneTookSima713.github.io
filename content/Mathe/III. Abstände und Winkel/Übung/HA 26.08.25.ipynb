### Nr. 5.
$$
\begin{align*}
g: \overrightarrow{ x } &= \begin{pmatrix}0\newline 5\newline 1\end{pmatrix} + t \begin{pmatrix}1\newline 2\newline 2\end{pmatrix} ~,\, \overrightarrow{ x } \text{in km, } t \text{ in }10 \text{ s} \\
h: \overrightarrow{ x } &= \begin{pmatrix}4\newline 9\newline 3\end{pmatrix} + t \begin{pmatrix}1\newline 1\newline 0\end{pmatrix} ~,\, \overrightarrow{ x } \text{in km, } t \text{ in }10 \text{ s}
\end{align*}
$$

$$
\begin{align*}
d(t) &= \left| \begin{pmatrix}0\newline 5\newline 1\end{pmatrix} + t \begin{pmatrix}1\newline 2\newline 2\end{pmatrix} - \begin{pmatrix}4\newline 9\newline 3\end{pmatrix} - t \begin{pmatrix}1\newline 1\newline 0\end{pmatrix} \right| \\
&= \left| \begin{pmatrix}-4\newline -4\newline -2\end{pmatrix} + t \begin{pmatrix}0\newline 1\newline 2\end{pmatrix} \right| \\
d'(t) &\overset != 0 \\
\end{align*}
$$

```python
t = symbols('t', real=True)
d = (Vec3(-4, -4, -2) + Vec3(0,1,2)*t).magnitude
sol = solve(d.diff(t), t)
(
    Eq(t, sol[0]),
    Eq(Function('d')(sol[0]), d.subs(t, sol[0])),
    Eq(Function('d')(sol[0]), d.subs(t, sol[0]).evalf())
)
```
