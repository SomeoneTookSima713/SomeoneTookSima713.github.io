#### a. Zeigen Sie, dass $g$ parallel zu $E$ verläuft.

$$
\begin{align*}
\begin{pmatrix}-1\newline -3\newline 2\end{pmatrix} \circ \begin{pmatrix}3\newline 1\newline 3\end{pmatrix} &= -3 - 3 + 6 \\
&= 0
\end{align*}
$$

#### b. Berechnen Sie den Abstand von $g$ zu $E$.

$$
\begin{align*}
\overrightarrow{ n } &= \begin{pmatrix}3\newline 1\newline 3\end{pmatrix} \\
\overrightarrow{ n_n } &= \frac{ \overrightarrow{ n } }{ \left| \overrightarrow{ n } \right| } \\
\text{Abst}(g; E) &= \left| \begin{pmatrix}1\newline 0\newline 4\end{pmatrix} \circ \overrightarrow{ n_n } - \begin{pmatrix}0\newline 4\newline 0\end{pmatrix} \circ \overrightarrow{ n_n } \right| \\
&= \frac{11\sqrt{19}}{19}
\end{align*}
$$

```python
n_n = Vec3(3,1,3).unit

Eq(Function("Abst")(Symbol("g"), Symbol("E")), abs(Vec3(1,0,4)*n_n - Vec3(0,4,0)*n_n).evalf())
```
