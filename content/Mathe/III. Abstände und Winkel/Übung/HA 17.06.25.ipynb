## LB. S. 357
### Nr. 13.
#### a.

$$
\begin{align*}
    \alpha_x &-~ \text{Winkel zwischen der Wand senkrecht zur Tiefenachse und der angrenzenden Dachfläche} \\
    \alpha_y &-~ \text{Winkel zwischen der Wand senkrecht zur Horizontalachse und der angrenzenden Dachfläche} \\
    \alpha_x &= 180\degree - \arctan\left( \frac{ 5 \text{ m} - 3 \text{ m} }{ 6 \text{ m} - 3 \text{ m} } \right) \\
    \alpha_y &= 180\degree - \arctan\left( \frac{ 3,25 \text{ m} }{ 6 \text{ m} - 3 \text{ m} } \right) \\
\end{align*}
$$

```python
ax, ay = symbols('alpha_x alpha_y', real=True)

ax_eval = pi - atan(Rational(2,3))
ay_eval = pi - atan(Rational(13,12))

( Eq(ax, deg(ax_eval).evalf()), Eq(ay, deg(ay_eval).evalf()) )
```

#### b.
$$
\begin{align*}
\overrightarrow{ n_x } &= \begin{pmatrix}\sin\alpha_x\newline 0\newline \cos\alpha_x\end{pmatrix} \\
\overrightarrow{ n_y } &= \begin{pmatrix}0\newline \sin\alpha_y\newline \cos\alpha_y\end{pmatrix} \\
\varphi &= \arccos\left( \overrightarrow{ n_x } \circ \overrightarrow{ n_y } \right) \\
&= \arccos(\cos\alpha_x \cdot \cos\alpha_y)
\end{align*}
$$

```python
varphi = symbols('varphi', real=True)

varphi_eval = acos( cos(ax_eval) * cos(ay_eval) )

Eq(varphi, deg(varphi_eval).evalf())
```
